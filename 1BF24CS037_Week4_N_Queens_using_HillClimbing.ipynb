```python
import random

def get_h_cost(state):
    """Calculate the number of attacking pairs of queens."""
    attacks = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Same column (not possible with state representation, but kept conceptually)
            if state[i] == state[j]:
                attacks += 1
            # Same diagonal
            elif abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_best_neighbor(state):
    """Find the neighbor with the lowest heuristic cost."""
    n = len(state)
    current_h = get_h_cost(state)
    best_state = list(state)
    best_h = current_h

    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbor_h = get_h_cost(neighbor)

                if neighbor_h < best_h:
                    best_h = neighbor_h
                    best_state = neighbor

    return best_state, best_h

def hill_climbing(n, max_restarts=100):
    """Solve N-Queens using Hill Climbing with random restarts and print search paths."""
    for restart in range(max_restarts):
        # Generate a random initial state where state[i] is the row index of queen in column i
        current_state = [random.randint(0, n - 1) for _ in range(n)]
        current_h = get_h_cost(current_state)

        print(f"\n--- Restart #{restart + 1} ---")
        print(f"Initial State: {current_state} | Heuristic Value (Clashes): {current_h}")

        step = 1
        while True:
            neighbor, neighbor_h = get_best_neighbor(current_state)

            if neighbor_h >= current_h:
                print(f"Stuck in local optimum at state: {current_state} | Heuristic Value: {current_h}")
                break

            current_state = neighbor
            current_h = neighbor_h
            print(f"  Step {step}: Move to state {current_state} | Heuristic Value (Clashes): {current_h}")
            step += 1

        if current_h == 0:
            print(f"\nGlobal Solution found after {restart} restarts!")
            return current_state

    print("\nFailed to find a solution within the maximum restarts limit.")
    return None

def print_board(state):
    """Utility function to print the chess board beautifully."""
    if state is None:
        return
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

# Take input for the value of N
try:
    n_input = int(input("Enter the value of N for N-Queens: "))
    if n_input < 4 and n_input != 1:
        print("N-Queens has no solution for N = 2 or 3.")
    else:
        solution = hill_climbing(n_input)
        if solution:
            print(f"\nOne of the solutions for {n_input}-Queens:")
            print("State representation (row position of queen in each column):", solution)
            print("\nBoard Layout:")
            print_board(solution)
except ValueError:
    print("Please enter a valid integer for N.")
```
